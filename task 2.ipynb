```python

import pandas as pd

# Membaca dataset training dan testing
df_train_fraud = pd.read_csv("klasifikasi_train.csv.gz")
df_test_fraud = pd.read_csv("klasifikasi_test.csv.gz")

print("Dataset berhasil dimuat!")

print("Ukuran training:", df_train_fraud.shape)
print("Ukuran testing :", df_test_fraud.shape)
```

```python

import pandas as pd
import numpy as np

# Memastikan kolom waktu bertipe datetime
df_train_fraud["waktu"] = pd.to_datetime(
    df_train_fraud["waktu"]
)
df_test_fraud["waktu"] = pd.to_datetime(
    df_test_fraud["waktu"]
)

# 1. Distribusi label fraud
print("DISTRIBUSI LABEL FRAUD")
print(df_train_fraud["is_fraud"].value_counts().sort_index())

print("\nPERSENTASE LABEL FRAUD")
print(
    (df_train_fraud["is_fraud"].value_counts(normalize=True)
     .sort_index() * 100).round(4)
)

# 2. Rentang waktu
print("\nRENTANG WAKTU DATA")
print("Training:")
print("Awal :", df_train_fraud["waktu"].min())
print("Akhir:", df_train_fraud["waktu"].max())

print("\nTesting:")
print("Awal :", df_test_fraud["waktu"].min())
print("Akhir:", df_test_fraud["waktu"].max())

# 3. Distribusi fraud per bulan
train_audit = df_train_fraud.copy()
train_audit["bulan"] = train_audit["waktu"].dt.to_period("M")

distribusi_bulanan = (
    train_audit.groupby("bulan")
    .agg(
        jumlah_transaksi=("is_fraud", "size"),
        jumlah_fraud=("is_fraud", "sum"),
        tingkat_fraud=("is_fraud", "mean")
    )
)

distribusi_bulanan["tingkat_fraud"] *= 100

print("\nDISTRIBUSI FRAUD PER BULAN")
display(distribusi_bulanan.round(3))

# 4. Missing value
print("\nMISSING VALUE TRAINING")
display(
    df_train_fraud.isna().sum()
    .sort_values(ascending=False)
    .head(15)
)

print("\nMISSING VALUE TESTING")
display(
    df_test_fraud.isna().sum()
    .sort_values(ascending=False)
    .head(15)
)

# 5. Duplikasi transaksi
print("\nDUPLIKASI ID TRANSAKSI")
print(
    "Training:",
    df_train_fraud["id_transaksi"].duplicated().sum()
)
print(
    "Testing:",
    df_test_fraud["id_transaksi"].duplicated().sum()
)
```

```python
# Menghitung jumlah fraud berdasarkan rancangan periode

periode = {
    "Training Jan-Mei": ("2025-01", "2025-05"),
    "Validation Juni": ("2025-06", "2025-06"),
    "Validation Juli-Agustus": ("2025-07", "2025-08")
}

for nama, (awal, akhir) in periode.items():

    data = df_train_fraud[
        (df_train_fraud["bulan"] >= awal) &
        (df_train_fraud["bulan"] <= akhir)
    ]

    print(f"\n{nama}")
    print("Jumlah transaksi:", len(data))
    print("Jumlah fraud:", data["is_fraud"].sum())
    print(
        "Tingkat fraud:",
        round(data["is_fraud"].mean() * 100, 3), "%"
    )
```

```python
# Pastikan data diurutkan berdasarkan waktu
df_train_fraud = df_train_fraud.sort_values("waktu").copy()

# Training Januari-Mei
df_train_model = df_train_fraud[
    df_train_fraud["waktu"] < "2025-06-01"
].copy()

# Validation 1: Juni
df_validasi_1 = df_train_fraud[
    (df_train_fraud["waktu"] >= "2025-06-01") &
    (df_train_fraud["waktu"] < "2025-07-01")
].copy()

# Validation 2: Juli-Agustus
df_validasi_2 = df_train_fraud[
    (df_train_fraud["waktu"] >= "2025-07-01") &
    (df_train_fraud["waktu"] < "2025-09-01")
].copy()

# Periksa ukuran dataset
print("TRAINING:", df_train_model.shape)
print("VALIDATION 1:", df_validasi_1.shape)
print("VALIDATION 2:", df_validasi_2.shape)
```

```python
# Kolom yang tidak digunakan sebagai fitur
kolom_dikeluarkan = [
    "id_transaksi",
    "id_pengguna",
    "id_merchant",
    "waktu",
    "bulan",
    "flag_review_manual",
    "is_fraud"
]

# Membuat fitur tambahan dari waktu transaksi
for df in [df_train_model, df_validasi_1, df_validasi_2]:
    df["bulan_transaksi"] = df["waktu"].dt.month
    df["hari_dalam_minggu"] = df["waktu"].dt.dayofweek

# Menentukan daftar fitur
fitur = [
    kolom for kolom in df_train_model.columns
    if kolom not in kolom_dikeluarkan
]

# Memisahkan fitur dan target
X_train = df_train_model[fitur].copy()
y_train = df_train_model["is_fraud"].copy()

X_validasi_1 = df_validasi_1[fitur].copy()
y_validasi_1 = df_validasi_1["is_fraud"].copy()

X_validasi_2 = df_validasi_2[fitur].copy()
y_validasi_2 = df_validasi_2["is_fraud"].copy()

# Identifikasi fitur kategorikal
fitur_kategorikal = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Ringkasan
print("JUMLAH FITUR:", len(fitur))
print("FITUR KATEGORIKAL:", fitur_kategorikal)
print("JUMLAH FITUR NUMERIK:", len(fitur) - len(fitur_kategorikal))

print("\nUKURAN DATA")
print("X_train:", X_train.shape)
print("X_validasi_1:", X_validasi_1.shape)
print("X_validasi_2:", X_validasi_2.shape)

print("\nDISTRIBUSI TARGET TRAINING")
print(y_train.value_counts())
```

```python
# Salin data agar data awal tetap aman
X_train_cb = X_train.copy()
X_validasi_1_cb = X_validasi_1.copy()

# Ubah fitur kategorikal menjadi string
for kolom in fitur_kategorikal:
    X_train_cb[kolom] = (
        X_train_cb[kolom]
        .fillna("__MISSING__")
        .astype(str)
    )

    X_validasi_1_cb[kolom] = (
        X_validasi_1_cb[kolom]
        .fillna("__MISSING__")
        .astype(str)
    )

print("Missing value kategorikal training:")
print(X_train_cb[fitur_kategorikal].isna().sum())

print("\nTipe data kategorikal:")
print(X_train_cb[fitur_kategorikal].dtypes)
```

```python
from catboost import CatBoostClassifier
from sklearn.metrics import average_precision_score

# Membuat model baseline
model_baseline = CatBoostClassifier(
    iterations=700,
    depth=6,
    learning_rate=0.05,
    loss_function="Logloss",
    eval_metric="PRAUC:type=Classic",
    random_seed=42,
    verbose=100,
    early_stopping_rounds=100,
    allow_writing_files=False
)

# Training model
model_baseline.fit(
    X_train_cb,
    y_train,
    cat_features=fitur_kategorikal,
    eval_set=(X_validasi_1_cb, y_validasi_1),
    use_best_model=True
)
```

```python
# Prediksi probabilitas fraud
prob_baseline = model_baseline.predict_proba(
    X_validasi_1_cb
)[:, 1]

# Menghitung PR-AUC
pr_auc_baseline = average_precision_score(
    y_validasi_1,
    prob_baseline
)

print("HASIL BASELINE CATBOOST")
print("Best iteration:", model_baseline.best_iteration_)
print("PR-AUC:", round(pr_auc_baseline, 5))
```

```python
from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

# Nominal transaksi pada validation Juni
nominal_validasi = df_validasi_1["nominal"].to_numpy()

# Label aktual
y_true = y_validasi_1.to_numpy()

# Fungsi menghitung biaya
def hitung_biaya(y_true, prediksi, nominal):
    fraud_lolos = (y_true == 1) & (prediksi == 0)
    jumlah_ditandai = prediksi.sum()

    kerugian_fraud = nominal[fraud_lolos].sum()
    biaya_review = jumlah_ditandai * 25000

    total_biaya = kerugian_fraud + biaya_review

    return {
        "fraud_lolos": fraud_lolos.sum(),
        "jumlah_ditandai": jumlah_ditandai,
        "kerugian_fraud": kerugian_fraud,
        "biaya_review": biaya_review,
        "total_biaya": total_biaya
    }

# Kondisi 1: Tidak ada transaksi ditandai
prediksi_none = np.zeros(len(y_true), dtype=int)

# Kondisi 2: Semua transaksi ditandai
prediksi_all = np.ones(len(y_true), dtype=int)

# Kondisi 3: Baseline CatBoost threshold 0.5
prediksi_baseline = (prob_baseline >= 0.5).astype(int)

# Menghitung biaya
hasil_none = hitung_biaya(
    y_true, prediksi_none, nominal_validasi
)

hasil_all = hitung_biaya(
    y_true, prediksi_all, nominal_validasi
)

hasil_baseline = hitung_biaya(
    y_true, prediksi_baseline, nominal_validasi
)

# Menampilkan hasil
hasil_biaya = pd.DataFrame(
    [hasil_none, hasil_all, hasil_baseline],
    index=["Flag None", "Flag All", "CatBoost Threshold 0.5"]
)

display(hasil_biaya)
```

```python
print("CONFUSION MATRIX")
print(confusion_matrix(y_true, prediksi_baseline))

print("\nPRECISION:")
print(round(precision_score(
    y_true, prediksi_baseline, zero_division=0
), 4))

print("\nRECALL:")
print(round(recall_score(
    y_true, prediksi_baseline, zero_division=0
), 4))

print("\nF1-SCORE:")
print(round(f1_score(
    y_true, prediksi_baseline, zero_division=0
), 4))
```

```python
import numpy as np
import pandas as pd

def cari_threshold_terbaik(y_true, probabilitas, nominal):
    daftar_hasil = []

    threshold_list = np.arange(0.01, 0.501, 0.005)

    for threshold in threshold_list:
        prediksi = (probabilitas >= threshold).astype(int)

        fraud_lolos = (y_true == 1) & (prediksi == 0)

        kerugian_fraud = nominal[fraud_lolos].sum()
        jumlah_ditandai = prediksi.sum()
        biaya_review = jumlah_ditandai * 25000

        total_biaya = kerugian_fraud + biaya_review

        daftar_hasil.append({
            "threshold": threshold,
            "fraud_lolos": fraud_lolos.sum(),
            "jumlah_ditandai": jumlah_ditandai,
            "kerugian_fraud": kerugian_fraud,
            "biaya_review": biaya_review,
            "total_biaya": total_biaya
        })

    hasil = pd.DataFrame(daftar_hasil)

    hasil_terbaik = hasil.loc[hasil["total_biaya"].idxmin()]

    return hasil, hasil_terbaik
```

```python
hasil_threshold_baseline, terbaik_baseline = cari_threshold_terbaik(
    y_true,
    prob_baseline,
    nominal_validasi
)

hasil_threshold_ros, terbaik_ros = cari_threshold_terbaik(
    y_true,
    prob_ros,
    nominal_validasi
)

print("THRESHOLD TERBAIK BASELINE")
display(terbaik_baseline.to_frame("Nilai"))

print("\nTHRESHOLD TERBAIK RANDOM OVERSAMPLING")
display(terbaik_ros.to_frame("Nilai"))
```

```python

def evaluasi_kelompok(data, kolom):
    hasil = (
        data.groupby(kolom)
        .apply(
            lambda g: pd.Series({
                "jumlah_transaksi": len(g),
                "jumlah_fraud": (g["is_fraud"] == 1).sum(),
                "fraud_terdeteksi": (
                    (g["is_fraud"] == 1) &
                    (g["prediksi_flag"] == 1)
                ).sum(),
                "fraud_lolos": (
                    (g["is_fraud"] == 1) &
                    (g["prediksi_flag"] == 0)
                ).sum(),
                "false_positive": (
                    (g["is_fraud"] == 0) &
                    (g["prediksi_flag"] == 1)
                ).sum(),
                "jumlah_ditandai": g["prediksi_flag"].sum(),
                "kerugian_fraud": g.loc[
                    (g["is_fraud"] == 1) &
                    (g["prediksi_flag"] == 0),
                    "nominal"
                ].sum(),
                "biaya_review": (
                    g["prediksi_flag"].sum() * 25000
                )
            })
        )
        .reset_index()
    )

    hasil["total_biaya"] = (
        hasil["kerugian_fraud"] + hasil["biaya_review"]
    )

    hasil["fraud_rate"] = (
        hasil["jumlah_fraud"] /
        hasil["jumlah_transaksi"] * 100
    )

    hasil["recall"] = (
        hasil["fraud_terdeteksi"] /
        hasil["jumlah_fraud"].replace(0, np.nan) * 100
    )

    return hasil


hasil_error_kanal = evaluasi_kelompok(
    df_error_agustus,
    "kanal"
)

print("EVALUASI ERROR BERDASARKAN KANAL")

display(
    hasil_error_kanal.sort_values(
        "total_biaya",
        ascending=False
    ).round(2)
)
```
